```python
import pandas as pd
import matplotlib.pyplot as plt
import os
df=pd.read_csv("ecommerce_customer_sales_intelligence.csv")
df.info()
df.columns
df.isna().sum() # MISSING VALUES
df.duplicated().sum()
df.drop_duplicates(inplace=True) # TO REMOVE DUPLICATES
df.isna().sum()
df["Quantity"].median()
df[df["Unit_Price"].isna()] # TO CHECK WHICH ONE IS MISSING
df[df["Discount_Pct"].isna()]
df["Unit_Price"].median()
df["Discount_Pct"].median()
df["Region"].mode()
df["Payment_Method"].mode()
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
df["Discount_Pct"] = df["Discount_Pct"].fillna(df["Discount_Pct"].median())
df["Region"] = df["Region"].fillna(df["Region"].mode()[0])
df["Payment_Method"] = df["Payment_Method"].fillna(df["Payment_Method"].mode()[0])
df.isna().sum() 
df[df["Unit_Price"].isna()]
df.loc[105, "Unit_Price"] = 15000
df.isna().sum() 
df["Date"]
df["Date"] = pd.to_datetime(df["Date"])
df.dtypes
df["Date"]
df["Payment_Method"].unique()
df["Region"].unique()
df["Category"].unique()
df["Product"].unique()
df.describe()
df["Sales"]=df["Quantity"]*df["Unit_Price"]
df["Sales"].head()
df["Discount_Amount"]=df["Sales"]*df["Discount_Pct"]/100
df["Discount_Amount"].head()
df[["Sales", "Discount_Pct", "Discount_Amount"]].head()
df["Net_Revenue"]=df["Sales"]-df["Discount_Pct"]
df["Net_Revenue"].head()
df["Total_Cost"]=df["Quantity"]*df["Cost_Price"]
df[["Quantity", "Cost_Price", "Total_Cost"]].head()
df["Profit"]=df["Net_Revenue"]-df["Total_Cost"]
df[["Net_Revenue", "Total_Cost", "Profit"]].head()
df["Profit_Margin"] = df["Profit"]/df["Net_Revenue"]*100
df[["Net_Revenue", "Profit", "Profit_Margin"]].head()
df.groupby("Product")["Net_Revenue"].sum()
df.groupby("Product")["Profit"].sum()
df.groupby("Category")["Net_Revenue"].sum()
df.groupby("Category")["Profit"].sum()
df.groupby("Region")["Net_Revenue"].sum()
df.groupby("Region")["Profit"].sum()
df.groupby("Customer_Type")["Net_Revenue"].sum()
df.groupby("Customer_Type")["Profit"].sum()
df.groupby(df["Date"].dt.month)["Net_Revenue"].sum()
df.groupby(df["Date"].dt.month)["Profit"].sum()
df.groupby("Customer_ID")["Net_Revenue"].sum().sort_values(ascending=False).head(5)
df.groupby("Customer_ID")["Profit"].sum().sort_values(ascending=False).head(5)
df.groupby("Discount_Pct")["Profit"].mean()
df.groupby("Product")["Profit"].sum().sort_values(ascending=False).head(5)
aov = df["Net_Revenue"].sum() / df["Order_ID"].nunique()
print(aov)
Average_Quantity_per_Order = df["Quantity"].sum() / df["Order_ID"].nunique()
print(Average_Quantity_per_Order)
df.groupby("Customer_ID")["Order_ID"].count()
monthly_revenue = df.groupby(df["Date"].dt.month)["Net_Revenue"].sum()

monthly_revenue.plot(kind="line", marker="o")

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Net Revenue")
plt.show()
df.to_csv("ecommerce_cleaned.csv", index=False)
```
